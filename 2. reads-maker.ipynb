```python
from repeat_adder import create_repeated_sequence, SEQ_SIZE, nt
import random
READ_LENGTH = 50
COVERAGE = 5
```

```python
repeat_seq = create_repeated_sequence('1025seq.txt', show_locations = True)
```

# GENERATE ARTIFICIAL READS FROM SEQ

```python
normal_reads = []
for _ in range(COVERAGE * 150):
    i = random.randint(0, SEQ_SIZE-READ_LENGTH)
    normal_reads.append("".join(repeat_seq[i:i+READ_LENGTH]))
print(normal_reads)
```

# ADD TO FASTQ FILE

```python
with open('repeat_seq.txt', 'w') as f:
    f.write("".join(repeat_seq))
```

```python
with open('reads.fa', 'w') as f:
    for i in range(len(normal_reads)):
        f.write(f">read{i}\n{normal_reads[i]}\n")
```

## Introduce Random Mutations

```python
MUTATION_RATE = 0.05
mutated_seq = repeat_seq.copy()
mutation_location = []
for i in range(SEQ_SIZE):
    if random.randint(1,100) <= MUTATION_RATE * 100:
        choices = {'A', 'T', 'C', 'G'} - {mutated_seq[i]}
        mutated_seq[i] = list(choices)[random.randint(0,2)]
        mutation_location.append(i)
```
