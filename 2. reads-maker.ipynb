```python
import random
READ_LENGTH = 50
SEQ_SIZE = 2484
COVERAGE = 10
READS_PER_PASS = 100
```

```python
with open('./sequence1.txt', 'r') as f:
    repeat_seq = f.read()
repeat_seq = repeat_seq[:-1]
SEQ_SIZE = len(repeat_seq)
repeat_seq
```

### GENERATE ARTIFICIAL READS FROM SEQ

```python
normal_reads = []
for _ in range(COVERAGE * READS_PER_PASS):
    i = random.randint(0, SEQ_SIZE-READ_LENGTH)
    normal_reads.append("".join(repeat_seq[i:i+READ_LENGTH]))
print(normal_reads)
```

### ADD TO FASTQ FILE

```python
#with open('a.txt', 'w') as f:
#    f.write("".join(repeat_seq))
```

```python
with open('reads1.fa', 'w') as f:
    for i in range(len(normal_reads)):
        f.write(f">read{i}\n{normal_reads[i]}\n")
```

## Introduce Random Mutations

```python
MUTATION_RATE = 0.05
mutated_seq = list(repeat_seq).copy()
mutation_location = []
for i in range(SEQ_SIZE):
    if random.randint(1,100) <= MUTATION_RATE * 100:
        choices = {'A', 'T', 'C', 'G'} - {mutated_seq[i]}
        mutated_seq[i] = list(choices)[random.randint(0,2)]
        mutation_location.append(i)
```

```python
print(f"{repeat_seq}\n=================================\n{"".join(mutated_seq)}")
```

```python
print(mutation_location)
```

```python
mutated_reads = []
for _ in range(COVERAGE * 150):
    i = random.randint(0, SEQ_SIZE-READ_LENGTH)
    mutated_reads.append("".join(mutated_seq[i:i+READ_LENGTH]))
print(mutated_reads)
```

```python
with open('mutate_reads1.fa', 'w') as f:
    for i in range(len(mutated_reads)):
        f.write(f">read{i}\n{mutated_reads[i]}\n")
```
